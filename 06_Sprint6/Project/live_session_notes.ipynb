```python
# Ver la distribución de ventas por plataforma
ventas_por_plataforma = datos.groupby('platform')['ventas_totales'].sum().sort_values(ascending=False).reset_index()
```

```python
ventas_por_plataforma
```

```python
# z-score: Determinar que tanto impacto tiene la plantaforma conrespecto al ingreso total
# Para saber si una plataforma es rentable o no, es necesario saber cuánto representan las ventas 
# de esa plataforma con respecto al promedio de todas la ventas
# ~ 46:50 :
# Elige varias plataformas potencialmente rentables
```

```python
ventas_por_plataforma['z-score'] = (ventas_por_plataforma['ventas_totales'] - ventas_por_plataforma['ventas_totales'].mean()) / ventas_por_plataforma['ventas_totales'].std()
```

```python
color = []

for score in ventas_por_plataforma['z-score']:
    if score < 0:
        color.append('red')
    else:
        color.append('green')

ventas_por_plataforma['color'] = color
```

```python
plt.hlines(
    y=ventas_por_plataforma['platform'],
    xmax=ventas_por_plataforma['z-score'],
    xmin=0,
    color=ventas_por_plataforma['color'],
    linewidth=10,
    alpha=0.5
)
```
